# House Price Prediction with Categorical Data

## Objective

This project predicts house prices based on area and town using Linear Regression.

Two methods are compared:
- Dummy variables with Pandas
- One-Hot Encoding with Scikit-learn

```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
```

## 1. Load the Dataset

```python
df = pd.read_csv('homepricessss.csv')
df
```

## 2. Dummy Variables with Pandas

```python
dummies = pd.get_dummies(df["town"], dtype=int)
dummies
```

```python
merged = pd.concat([df, dummies], axis="columns")
merged
```

```python
final = merged.drop(["town"], axis="columns")
final
```

```python
final = final.drop(['west windsor'], axis='columns')
final 
```

```python
#X = final.drop('price', axis='columns')
X = final[['area','monroe township','robinsville']]
X
```

```python
y=final.price
```

```python

model = LinearRegression()
```

```python
model.fit(X,y)
```

```python
model.coef_
```

```python
model.intercept_
```

```python
model.predict(X) 
```

```python
model.score(X,y)
```

```python
new_house = pd.DataFrame({
    "area": [3400],
    "monroe township": [0],
    "robinsville": [0]
})

model.predict(new_house)
```

## 3. One-Hot Encoding with Scikit-learn

```python
X_ohe = df[["town", "area"]].copy()
X_ohe
```

```python
y = df["price"]
y
```

```python
ct = ColumnTransformer(transformers=[("town",OneHotEncoder(drop="first"),["town"])],remainder="passthrough")
```

```python
X_encoded = ct.fit_transform(X_ohe)
X_encoded
```

```python
ohe_model = LinearRegression()
ohe_model.fit(X_encoded, y)
```

```python
ohe_model.coef_
```

```python
ohe_model.score(X_encoded, y)
```
