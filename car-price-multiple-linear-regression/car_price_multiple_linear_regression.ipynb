# Car Price Prediction with Multiple Linear Regression

## Objective

This project predicts car selling prices using:
- Mileage
- Age of the car

The dataset is split into training and testing sets to evaluate the regression model.

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("carprices.csv")
df.head()
```

```python
plt.scatter(df['Mileage'],df['Sell Price($)'])
```

```python
plt.scatter(df['Age(yrs)'],df['Sell Price($)'])
```

```python
X = df[['Mileage','Age(yrs)']]
y=df[['Sell Price($)']]
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.3)
```

```python
X_train
```

```python
X_test
```

```python
y_train
```

```python
y_test
```

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
y_pred
```

```python
model.score(X_test,y_test)
```

```python
results = pd.DataFrame({
    "Actual Price": y_test.values.ravel(),
    "Predicted Price": y_pred.ravel()})
results
```

## Conclusion

A Multiple Linear Regression model was trained to predict car selling prices based on mileage and age.

The dataset was divided into training and testing sets to evaluate the model on unseen data.
